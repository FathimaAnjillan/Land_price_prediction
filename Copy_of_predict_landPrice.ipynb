```python

import pandas as pd
df=pd.read_csv('land_prices.csv')
print(df.head())
```

```python
#cleaning
null_values=df.null.sum()
print(null_values)
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
#FEATURE ENGINEERING
#finding price per cent
df["price_per_cent"]=df['price_lakhs']/ df['land_area_cents']
```

```python
#print(df.columns)
#extracting features
numeric_features=['property_id',
       'longitude', 'land_area_cents', 'distance_to_school_km',
       'distance_to_airport_km', 'distance_to_railway_station_km',
       'distance_to_hospital_km', 'distance_to_medical_college_km',
       'distance_to_bus_stop_km', 'distance_to_market_km', 'price_lakhs',
       'price_per_cent']
required_features=[
       'longitude', 'land_area_cents', 'distance_to_school_km',
       'distance_to_airport_km', 'distance_to_railway_station_km',
       'distance_to_hospital_km', 'distance_to_medical_college_km',
       'distance_to_bus_stop_km', 'distance_to_market_km']
#Encoding
df_processed=pd.get_dummies(df,columns=['location_name','taluk','village','land_type'],drop_first=True)
print(df_processed.head())

```

```python
dummy_features=[]
for column in df_processed:
  if column.startswith('location_name')or column.startswith('taluk')or column.startswith('village')or column.startswith('land_type'):
    dummy_features.append(column)
print(dummy_features)
```

```python
features=required_features+dummy_features#festures for the model
target=df['price_per_cent']
#ml part
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
#testing and trainign data split
x=df_processed[features]
y=target
#print(y)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
#create model
model=RandomForestRegressor(n_estimators=100,random_state=42)
#training
model.fit(x_train,y_train)
#testing
predict_result=model.predict(x_test)
#print(predict_result)
df_comparison=pd.DataFrame({'actual':y_test,'predicted':predict_result})
print(df_comparison.head())


```
